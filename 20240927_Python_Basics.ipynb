```python
# Lambda function = a small anonymous function for a one time use (throw away function)
#                   they take any numnber of arguments, but only have one expresiion
#                   helps keep the namespace clean and useful with higher order functions
#                   'sort()', 'map()', 'filter()', 'reduce()'
#                    lambda parameters: expression

double = lambda x: x * 2
add = lambda x, y: x + y
max_value = lambda x, y: x if x > y else y
min_value = lambda x, y: x if x < y else y
full_name = lambda first_name, last_name: first_name + " " + last_name
is_even = lambda x: x % 2 == 0
is_odd = lambda x: x % 2 != 0
age_check = lambda age: True if age >= 18 else False


```

```python
import random

# Lambda functions
double = lambda x: x * 2
add = lambda x, y: x + y
max_value = lambda x, y: x if x > y else y
min_value = lambda x, y: x if x < y else y
full_name = lambda first_name, last_name: first_name + " " + last_name
is_even = lambda x: x % 2 == 0
is_odd = lambda x: x % 2 != 0
age_check = lambda age: True if age >= 18 else False

# Variables
x = random.randint(1, 100)
y = random.randint(1, 100)
first_name = "John"
last_name = "Doe"
age = random.randint(0, 100)

# Printing results
print("Double:", double(x))
print("Add:", add(x, y))
print("Max value:", max_value(x, y))
print("Min value:", min_value(x, y))
print("Full name:", full_name(first_name, last_name))
print("Is even:", is_even(x))
print("Is odd:", is_odd(x))
print("Age check (18+):", age_check(age))
```

```python
from scipy.stats import norm

# Parameters
mean = 502  # Mean score
std_dev = 115  # Standard deviation
lower_bound = 400  # Lower score threshold
upper_bound = 590  # Upper score threshold

# Calculate cumulative probabilities
prob_upper = norm.cdf(upper_bound, loc=mean, scale=std_dev)  # P(X <= 590)
prob_lower = norm.cdf(lower_bound, loc=mean, scale=std_dev)  # P(X <= 400)

# Calculate the probability between 400 and 590
probability = prob_upper - prob_lower

# Convert to percentage
percentage = probability * 100

print(prob_upper)
print(prob_lower)
print(f"Percentage of students scoring between 400 and 590: {percentage:.2f}%")
```

```python
from scipy.stats import binom

# Parameters
n = 7  # Number of questions
p = 0.75  # Probability of guessing correctly
k = 3  # Maximum number of correct answers

# Calculate the cumulative probability using the CDF
probability = binom.cdf(k, n, p)

print(f"Probability of answering at most 3 questions correctly: {probability:.4f}")
```

```python
import numpy as np
from scipy import stats

# Given data
confidence_coef = 0.95
df = 100 - 1  # Degrees of freedom
loc = 43      # Sample mean
scale = 3.6 / np.sqrt(100)  # Standard error

# Calculate the confidence interval
confidence_interval = stats.t.interval(confidence_coef, df=df, loc=loc, scale=scale)

# Round the confidence interval to two decimal places
confidence_interval_rounded = (round(confidence_interval[0], 2), round(confidence_interval[1], 2))

print("95% Confidence Interval (rounded to two decimal places):", confidence_interval_rounded)
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Create the contingency table
# [rookie_crashed, rookie_not_crashed]
# [experienced_crashed, experienced_not_crashed]

contingency_table = np.array([
    [4, 2],  # 4 rookies crashed, 2 rookies didn't crash
    [2, 12]  # 2 experienced drivers crashed, 12 experienced drivers didn't crash
])

# Perform Fisher's exact test
odds_ratio, p_value = stats.fisher_exact(contingency_table, alternative='greater')

# Print the results
print(f"Contingency Table:")
print(f"               | Crashed | Not Crashed | Total")
print(f"---------------+---------+-------------+------")
print(f"Rookies        |    4    |      2      |   6  ")
print(f"Experienced    |    2    |     12      |  14  ")
print(f"---------------+---------+-------------+------")
print(f"Total          |    6    |     14      |  20  ")
print("\nFisher's Exact Test Results:")
print(f"p-value: {p_value:.4f}")
print(f"Odds Ratio: {odds_ratio:.2f}")

if p_value < 0.05:
    print("\nThe result is statistically significant at α = 0.05.")
    print("We can reject the null hypothesis that rookies and experienced drivers crash at the same rate.")
    print("The data suggests that rookies were more likely to crash in this race.")
else:
    print("\nThe result is not statistically significant at α = 0.05.")
    print("We fail to reject the null hypothesis.")

# Calculate crash rates as percentages
rookie_crash_rate = (4 / 6) * 100
experienced_crash_rate = (2 / 14) * 100
print(f"\nCrash rates:")
print(f"Rookies: {rookie_crash_rate:.1f}%")
print(f"Experienced drivers: {experienced_crash_rate:.1f}%")

# Create a visualization of the results
plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")

# Create the data for plotting
categories = ['Rookies', 'Experienced']
crashed = [4, 2]
not_crashed = [2, 12]

# Create a bar plot
bar_width = 0.35
x = np.arange(len(categories))

plt.bar(x - bar_width/2, crashed, bar_width, label='Crashed', color='red', alpha=0.7)
plt.bar(x + bar_width/2, not_crashed, bar_width, label='Not Crashed', color='green', alpha=0.7)

# Add percentage labels on top of the bars
for i, v in enumerate(crashed):
    plt.text(i - bar_width/2, v + 0.1, f'{v/sum([v, not_crashed[i]])*100:.1f}%',
             ha='center', va='bottom', fontweight='bold')
for i, v in enumerate(not_crashed):
    plt.text(i + bar_width/2, v + 0.1, f'{v/sum([crashed[i], v])*100:.1f}%',
             ha='center', va='bottom', fontweight='bold')

plt.xlabel('Driver Experience')
plt.ylabel('Number of Drivers')
plt.title('F1 Australia 2025: Driver Crashes by Experience\n'
          f'p-value = {p_value:.4f}, Odds Ratio = {odds_ratio:.2f}')
plt.xticks(x, categories)
plt.legend()
plt.tight_layout()

# Add a text box with the test results
textstr = f"Fisher's Exact Test\np-value: {p_value:.4f}\nOdds Ratio: {odds_ratio:.2f}"
props = dict(boxstyle='round', facecolor='wheat', alpha=0.5)
plt.gcf().text(0.15, 0.15, textstr, fontsize=10, bbox=props)

# Show the plot
plt.show()
```
